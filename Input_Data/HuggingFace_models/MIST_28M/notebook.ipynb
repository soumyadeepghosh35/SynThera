```python
from transformers import AutoModel
from smirk import SmirkTokenizerFast

# Load the model
model = AutoModel.from_pretrained(
    'mist-models/mist-28M-ti624ev1',
    trust_remote_code=True
)

# Make predictions
smiles_batch = [
    "CCO",           # Ethanol
    "CC(=O)O",       # Acetic acid
    "C1=CC=CC=C1"       # Benzene
]
results = model.predict(smiles_batch)
results
```
